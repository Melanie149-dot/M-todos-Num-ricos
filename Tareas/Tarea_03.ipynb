````python
import math

def cortardigitos(x):                     
    if x == 0: 
        return 0.0
    exp = math.floor(math.log10(abs(x)))        
    factor = 10 ** (2 - exp)                    
    return math.trunc(x * factor) / factor      

def sumacorte(serie, orden='asc'):                                
    total = 0.0                                                        
    datos = serie if orden == 'asc' else list(reversed(serie))         
    for x in datos:                                                    
        total = cortardigitos(total + cortardigitos(x))
    return total                                                       

def mostrar_errores(valor_real, valor_aprox):
    error_abs = abs(valor_real - valor_aprox)
    error_rel = error_abs / abs(valor_real)
    print(f"Error absoluto: |{valor_real:.9f} - {valor_aprox:.9f}| = {error_abs:.9f}")
    print(f"Error relativo: |({valor_real:.9f} - {valor_aprox:.9f}) / {valor_real:.9f}| = {error_rel:.9f}\n")

sumatoria_a = [1 / (i**2) for i in range(1, 11)]
sumatoria_b = [1 / (i**3) for i in range(1, 11)]

a_asc = sumacorte(sumatoria_a, orden='asc')
a_desc = sumacorte(sumatoria_a, orden='desc')
b_asc = sumacorte(sumatoria_b, orden='asc')
b_desc = sumacorte(sumatoria_b, orden='desc')

suma_real_a = sum(1 / (i**2) for i in range(1, 11)) 
suma_real_b = sum(1 / (i**3) for i in range(1, 11))

print("Para el literal a\n")
print(f"Sumatoria sin corte: {suma_real_a:.9f}")
print(f"Suma ascendente con corte: {a_asc}")
mostrar_errores(suma_real_a, a_asc)
print(f"Suma descendente con corte: {a_desc}")
mostrar_errores(suma_real_a, a_desc)

print("Para el literal b\n")
print(f"Sumatoria sin corte: {suma_real_b:.9f}")
print(f"Suma ascendente con corte: {b_asc}")
mostrar_errores(suma_real_b, b_asc)
print(f"Suma descendente con corte: {b_desc}")
mostrar_errores(suma_real_b, b_desc)
````

# Ejercicio 2

La serie de Maclaurin para la función arcotangente converge para −1<𝑥≤1 y está dada por:

$\arctan x = \lim_{n \to \infty} P_n(x) \equiv \lim_{n \to \infty} \sum_{i=1}^{n} (-1)^{i+1} \frac{x^{2i-1}}{2i - 1}$


a. Utilice el hecho de que tan 𝜋 4⁄ =1 para determinar el número n de términos de la serie que se necesita 
sumar para garantizar que |4𝑃𝑛(1)−𝜋|<10−3 

b. El lenguaje de programación C++ requiere que el valor de 𝜋 se encuentre dentro de 10−10. ¿Cuántos 
términos de la serie se necesitarían sumar para obtener este grado de precisión? 

````python
import math

def calcularnecesario(tolerancia_error_pi):          
    # n se calcula basándose en la serie de Leibniz para pi/4
    n = math.ceil((4 / tolerancia_error_pi - 1) / 2)    
    return int(n)                                     

def calcularpi(n_terminos):                 
    suma_parcial_pi_4 = 0.0                             

    for i in range(1, n_terminos + 1):                  
        denominador = 2 * i - 1                         
        signo = (-1)**(i + 1)                           
        termino = signo / denominador                   
        suma_parcial_pi_4 += termino                    
        
    return 4 * suma_parcial_pi_4                      

pi_referencia = 3.141592653589793  

# --- Literal a ---
print("Literal a")
tolerancia_a = 10**-3  # PRIMERO definimos la variable
n_a = calcularnecesario(tolerancia_a) # LUEGO la usamos
pi_aprox_a = calcularpi(n_a)                    
error_a = abs(pi_aprox_a - pi_referencia)

print(f"Número de términos requeridos (n): {n_a}")
print(f"Pi aproximado con {n_a} términos: {pi_aprox_a}")
print(f"Error absoluto: {error_a:.6f} (< {tolerancia_a})")

# --- Literal b ---
print("\nLiteral b")
tolerancia_b = 10**-10  
n_b = calcularnecesario(tolerancia_b)
print(f"Número de términos requeridos (n): {n_b:,}")
````

# Ejercicio 5


a.- ¿Cuántas multiplicaciones y sumas se requieren para determinar una suma de la forma 

$$\sum_{i=1}^{n} \sum_{j=1}^{i} a_ib_j$$

b.- Modifique la suma en la parte a) a un formato equivalente que reduzca el número de cálculos. 

````python
def contar_operacion(n):
    multiplicacion = n * (n + 1) // 2
    suma = n * (n - 1) // 2
    return multiplicacion, suma

n = 9
multiplicacion, suma = contar_operacion(n)
print("\nLiteral a\n")
print(f"Para n = {n}, se necesitan {multiplicacion} multiplicaciones y {suma} sumas.")

def suma_optimizada(a, b):
    n = len(a)
    suma_acumulada = [0] * n
    suma_acumulada[-1] = a[-1]

    for i in range(n - 2, -1, -1):                              
        suma_acumulada[i] = suma_acumulada[i + 1] + a[i]

    resultado = 0
    for j in range(n):                              
        resultado += b[j] * suma_acumulada[j]

    return resultado

a = [2, 4, 6, 8, 10]
b = [1, 3, 5, 7, 9]
resultado = suma_optimizada(a, b)
print("\nLiteral b\n")  
print("Resultado de la suma optimizada:", resultado)
````
