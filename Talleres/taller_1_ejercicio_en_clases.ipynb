2. Definimos la funcion f (x)

```python
def f (x):
    return (1/4) * (x**3 + 3*x**2 - 6*x - 8)
```

3. Graficamos la funcion

```python
x = np.linspace (-6, 4, 1000)
y = f (x)
plt.plot (x, y)
plt.axhline (0, color = 'gray', linestyle = '--') # Agrega una linea horizontal para marcar el eje x
plt.title ('Grafico de la funcion f(x)') # Titulo del grafico
plt.xlabel ('x')
plt.ylabel ('f(x)')
plt.grid (True) # Agrega una cuadrícula al gráfico para mejorar la legibilidad
plt.show ()
```

4.  Crea modelo de biseccion 

```python
def biseccion (a, b, interacciones):
    for i in range (interacciones):
        xr = (a + b) / 2

        print (f"Interacion {i + 1}")
        print (f"a = {a:.6f}")
        print (f"b = {b:.6f}")
        print (f"xr = {xr:.6f}")
        print (f"f(xr) = {f(xr):.6f}")

        if f(a) * f (xr) < 0:
            b = xr
        else:
            a = xr  
    return xr
```
