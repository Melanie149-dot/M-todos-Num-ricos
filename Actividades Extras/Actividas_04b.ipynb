# Ejercicio 1

¿Cuál es la diferencia entre FP32 vs TP32?

La diferencia fundamental es que **FP32** es un formato de alta fidelidad que prioriza la precisión absoluta, mientras que **TF32** es un formato de procesamiento acelerado diseñado para maximizar la velocidad en tareas de inteligencia artificial. FP32 utiliza 23 bits para la mantisa, lo que permite una enorme exactitud decimal en cálculos científicos o matemáticos complejos. En cambio, TF32 reduce esa precisión a solo 10 bits (equivalente a FP16) para agilizar las operaciones en los núcleos de la GPU, pero mantiene el mismo rango dinámico de 8 bits en el exponente para evitar errores de desbordamiento de datos.

En la práctica, TF32 actúa como un formato intermedio que permite a las tarjetas gráficas modernas procesar información de FP32 de forma mucho más eficiente sin necesidad de que el usuario cambie su código. Al sacrificar la precisión de los decimales menos significativos —que suelen ser irrelevantes para el entrenamiento de redes neuronales—, se logra multiplicar el rendimiento computacional hasta diez veces en comparación con el estándar FP32 tradicional, manteniendo la estabilidad necesaria para que el modelo aprenda correctamente.

# Ejercicio 2

¿Qué representaciones de datos soportan estas GPUs (FP64, FP32, INT8)?

Las GPUs modernas están diseñadas para manejar una jerarquía de formatos que se adaptan a la necesidad de cada tarea. El formato **FP64** se reserva para la computación científica de alta precisión, donde se requieren cálculos exactos sin margen de error, aunque es el que más recursos consume y el más lento de procesar. Por otro lado, el **FP32** es el estándar equilibrado para gráficos de videojuegos y tareas generales, ofreciendo una precisión suficiente para la mayoría de las aplicaciones sin sacrificar demasiado rendimiento.

En el extremo opuesto se encuentra el formato **INT8**, que utiliza números enteros en lugar de decimales complejos para realizar operaciones masivas a una velocidad altísima. Este formato es fundamental para la inferencia de modelos de inteligencia artificial, ya que permite que dispositivos con menor potencia ejecuten procesos rápidamente al reducir drásticamente el uso de memoria. Además de estos, las arquitecturas actuales han incorporado formatos intermedios como **FP16** y **BF16**, que permiten entrenar redes neuronales de forma eficiente, logrando un balance ideal entre la velocidad de los enteros y el rango dinámico de los números decimales.

# Ejercicio 3

¿Por qué la nueva arquitectura prefiere representaciones numéricas con menor precisión?

La transición hacia precisiones menores en las nuevas arquitecturas se debe principalmente a un compromiso entre la **eficiencia de memoria y la velocidad de procesamiento**. En el campo de la inteligencia artificial y el aprendizaje profundo, se ha descubierto que las redes neuronales son inherentemente resistentes al ruido; esto significa que no necesitan la precisión de 23 bits de la mantisa de un FP32 para aprender patrones o realizar predicciones precisas. Al reducir el tamaño de los datos a formatos como FP8 o incluso INT8, se reduce drásticamente el ancho de banda necesario para mover información entre la memoria y los núcleos de la GPU, lo que permite entrenar modelos mucho más grandes en el mismo hardware y disminuir el consumo energético.

Además, el uso de menor precisión permite a los fabricantes de chips diseñar unidades de cálculo llamadas **Tensor Cores** que pueden realizar muchas más operaciones matemáticas por ciclo de reloj. Mientras que un núcleo tradicional procesa una operación compleja en FP64, un Tensor Core puede procesar decenas de operaciones en formatos más pequeños simultáneamente. Esto es vital para la era actual de los grandes modelos de lenguaje (LLM), donde la prioridad no es la exactitud decimal extrema de un solo número, sino la capacidad de procesar billones de parámetros y operaciones de matrices de forma casi instantánea para que la respuesta de la IA sea fluida y económicamente viable.
